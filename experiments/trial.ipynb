```python
# lets get some mandelbrot drawing 

import shap
import numpy as np
import matplotlib.pyplot as plt

xlim = (-2.0, 1.0)
ylim = (-1.5, 1.5)
window_with = 0.2
window_center_x = 0.46
window_center_y = 0.34
xlim = np.add(np.repeat((window_center_x), 2), np.multiply((-0.1, 0.1), window_with))
ylim = np.add(np.repeat((window_center_y), 2), np.multiply((-0.1, 0.1), window_with))

def mandelbrot(h, w, maxit=20):

    y, x = np.ogrid[ylim[0]:ylim[1]:h*1j, xlim[0]:xlim[1]:w*1j]
    c = x + 1j * y
    z = c
    divtime = maxit + np.zeros(z.shape, dtype=int)

    for i in range(maxit):
        z = z**2 + c
        diverge = z*np.conj(z) > 4  # who is diverging
        div_now = diverge & (divtime == maxit)  # who is diverging now
        divtime[div_now] = i  # note when
        z[diverge] = 2  # avoid diverging too much

    return np.flip(divtime, axis=0)
    
print(xlim, ylim)

plt.imshow(mandelbrot(200, 200),
 extent=(xlim[0], xlim[1], ylim[0], ylim[1]))
plt.show()
```
